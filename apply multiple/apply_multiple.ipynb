```python
import pandas as pd
import numpy as np
```

```python
stuff = {
    'Corporation':['Apple', 'Google', 'Meta', 'Apple', 'Google','Meta'],
    'Employees':['John', 'April', 'Wes', 'Beth', 'Justin', 'Steph'],
    'Salary':[200, 220, 190, 130, 120, 150],
}
df = pd.DataFrame(stuff)
df
```

**create function**

```python
def times1000(x):
    return format(x * 1000, ',d')
df['Salary'].apply(times1000)
```

**Append to the current dataframe**

```python
df["Salary"] = df["Salary"].apply(times1000)
df
```

**Fnction on multiple column**

```python
def namer(x):
    return 'RAMIM: ' + x

# apply to two columns
df[['Corporation', 'Employees']].apply(namer)
```

**Append to the current DataFrame**

```python
df[['Corporation', 'Employees']] = df[['Corporation', 'Employees']].apply(namer)
df
```

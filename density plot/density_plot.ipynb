```python
import pandas as pd
import numpy as np
from numpy.random import randn
%matplotlib inline
```

```python
df = pd.DataFrame(randn(500,4), columns=["Mon", "Tues", "Wed", "Thurs"])
df
```

```python
df.plot(kind='kde')
```

```python
df.plot.kde()
```

```python
df.plot.density()
```

```python
df.plot(kind='density')
```

```python
df['Mon'].plot.kde(alpha=0.4, lw=4)
```

```python
df['Mon'].plot.kde(alpha=0.4, lw=4, grid=True)
```

```python
df['Mon'].plot.kde(alpha=0.4, lw=4, grid=True, title='My KDE Plot')
```

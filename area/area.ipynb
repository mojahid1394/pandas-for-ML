```python
import pandas as pd
import numpy as np
from numpy.random import randn
%matplotlib inline
```

```python
# Create Dataframe
df = pd.DataFrame(randn(25,4), columns=["Mon", "Tues", "Wed", "Thu"])
df
```

**area plot**

```python
df.abs().plot(kind='area')
```

**Shading**

```python
df.abs().plot(kind='area', alpha = 0.3)
```

**Title**

```python
df.abs().plot(kind='area', alpha = 0.3, title='My Area Plot')
```

**Legend**

```python
df.abs().plot(kind='area', title='My Area Plot', alpha = 0.3, legend=False)
```

**Plot only one column**

```python
df['Mon'].abs().plot(kind='area', alpha = 0.3, title='My Area Plot', legend = False)
```

**Plot two columns**

```python
df[['Mon','Thu']].abs().plot(kind='area', alpha = 0.3, title='My Area Plot', legend = False)
```

**Add Table**

```python
df.abs().plot(kind='area', alpha = 0.3, title='My Area Plot', legend=False, table=True)
```

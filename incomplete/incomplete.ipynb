```python
import pandas as pd
import numpy as np
```

```python
stuff = {'A':[1,2,3], 'B':[4,np.nan,6], 'C':[7,8,9], 'D':[10,11,12]}
df = pd.DataFrame(stuff)
df
```

drop rows with null data

```python
df.dropna(axis = 0)
```

drop column with null data

```python
df.dropna(axis = 1)
```

```python
# not permanent, inplace = false
df 
```

more then one? set threshold

```python
df.dropna(thresh=2, axis = 1)
```

replace with fillna()

```python
df.fillna(value=40)
```

using math function

```python
#avg
df.fillna(value=df['B'].mean())
```

```python
#max
df.fillna(value=df['B'].max())
```

```python
#min
df.fillna(value=df['B'].min())
```

```python
#sum
df.fillna(value=df['B'].sum())
```

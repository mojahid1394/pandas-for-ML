```python
import pandas as pd
import numpy as np
from numpy.random import randn
%matplotlib inline
```

```python
# create a dataframe with some random number
df = pd.DataFrame(randn(100, 4), columns=['Mon', 'Tues', 'Wed', 'Thus'])
df
```

**Histogram**

```python
df['Wed'].hist()
```

**Histogram with Smaller bins**

```python
df['Wed'].hist(bins=20)
```

**without grid bars**

```python
df['Wed'].plot(kind='hist', bins = 20)
```

```python
df['Mon'].hist(bins=20, grid=False, legend=True)
```

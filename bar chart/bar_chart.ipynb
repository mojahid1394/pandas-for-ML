```python
import pandas as pd
import numpy as np
from numpy.random import randn
%matplotlib inline 
```

```python
# create a dataframe
df = pd.DataFrame(randn(25, 4), columns=["Mon", "Tues", "Wed", "Thu"])
df
```

**bar plot**

```python
df.abs().plot(kind='bar')
```

**bar plot Stacked**

```python
df.abs().plot(kind='bar', stacked=True)
```

**Shading**

```python
df.abs().plot(kind='bar', stacked=True, alpha = 0.8)
```

**Title**

```python
df.abs().plot(kind='bar', stacked=True, alpha=0.8, title='My Bar Chart')
```

**Legend**

```python
df.abs().plot(kind='bar', alpha=0.8, stacked=True, title='My Bar Chart', legend=False)
```

**SECOND WAY**

```python
df.abs().plot.bar()
```

```python
df.abs().plot.bar(stacked = True)
```

```python
df.abs().plot.bar(stacked = True, alpha=0.8)
```

```python
df.abs().plot.bar(stacked=True, alpha=0.8, title='Bar Chart!!')
```

```python
df.abs().plot.bar(stacked=True, alpha=0.8, title='Bar Chart!!', legend=False)
```

```python
df.abs().plot.bar(stacked=True, alpha=0.8, title='Bar Chart!!', legend=False, grid=True)
```

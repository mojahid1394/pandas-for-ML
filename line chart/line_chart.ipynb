```python
import pandas as pd
import numpy as np
from numpy.random import randn
%matplotlib inline
```

```python
# create a dataframe
df = pd.DataFrame(randn(25, 4), columns=["Mon", "Tues", "Wed", "Thu"])
df
```

**Line plot**

```python
df.plot(kind='line')
```

```python
df.abs().plot(kind='line')
```

**Line Width**

```python
df.abs().plot(kind='line', lw=5)
```

**line on single column**

```python
df['Mon'].abs().plot(kind='line', lw=5, legend=True)
```

**line more than one column**

```python
df[['Mon','Tues']].abs().plot(kind='line', lw=5)
```

**Size**

```python
df[['Mon', 'Tues']].abs().plot(kind='line', lw=5, figsize=(10, 5))
```

**Title**

```python
df[['Mon', 'Tues']].abs().plot(kind='line', lw=5, figsize=(10, 5), title='My Line Chart')
```

**Shading**

```python
df[['Mon', 'Tues']].abs().plot(kind='line', lw=5, figsize=(10, 5), title='My Line Chart', alpha=0.7)
```

**Legend**

```python
df[['Mon', 'Tues']].abs().plot(kind='line', lw=5, figsize=(10, 5), title='My Line Chart', alpha=0.7, legend=False)
```

**SECOND WAY**

```python
df[['Mon', 'Tues']].abs().plot.line()
```

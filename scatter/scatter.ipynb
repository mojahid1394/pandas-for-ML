```python
import pandas as pd
import numpy as np
from numpy.random import randn
%matplotlib inline
```

```python
df = pd.DataFrame(randn(500,4), columns=["Mon", "Tues", "Wed", "Thurs"])
df
```

**Scatter plot**

```python
df.plot(kind='scatter', x='Mon', y='Tues', c='Wed')
```

```python
# Colors: https://matplotlib.org/stable/tutorials/colors/colormaps.html
df.plot(kind='scatter', x='Mon', y='Tues', c='Wed', cmap='magma')
```

**Size**

```python
df.plot(kind='scatter', x='Mon', y='Tues', s=df['Wed']*100)
```

**By size easier to read with alpha**

```python
df.plot(kind='scatter', x='Mon', y='Tues', s=df['Wed']*100, alpha=0.8)
```

**Title**

```python
df.plot(kind='scatter', x='Mon', y='Tues', s=df['Wed']*100, alpha=0.8, title='My Scatter Plot')
```

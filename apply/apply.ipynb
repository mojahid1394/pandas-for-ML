```python
import pandas as pd
import numpy as np
```

```python
stuff = {
    'Corporation':['Apple', 'Google', 'Meta', 'Apple', 'Google','Meta'],
    'Employees':['John', 'April', 'Wes', 'Beth', 'Justin', 'Steph'],
    'Salary':[200, 220, 190, 130, 120, 150]}
df = pd.DataFrame(stuff)
df
```

create function

```python
def times1000(x):
    return format(x * 1000, ',d')
```

using apply function

```python
df['Salary'].apply(times1000)
```

```python
pd.DataFrame(df['Salary'].apply(times1000))
```

add Elder as Last name

```python
def namer(x):
    if x == 'John': 
        return 'John Elder'
    else:
        return x
```

use apply function

```python
df['Employees'].apply(namer)
```

```python
pd.DataFrame(df['Employees'].apply(namer))
```

Use Lambda

```python
df['Salary'].apply(lambda x: format(x * 1000, ',d'))
```

```python
pd.DataFrame(df['Salary'].apply(lambda x: format(x * 1000, ',d')))
```
